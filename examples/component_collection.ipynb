```python
import numpy as np

from easydynamics.sample_model import Gaussian
from easydynamics.sample_model import Lorentzian
from easydynamics.sample_model import DampedHarmonicOscillator
from easydynamics.sample_model import Polynomial

from easydynamics.sample_model import ComponentCollection


import matplotlib.pyplot as plt


%matplotlib widget
```

```python
from scipy.integrate import simpson

model = ComponentCollection(display_name="TestComponentCollection")
component1 = Gaussian(
    display_name="TestGaussian1",
    area=1.0,
    center=0.0,
    width=1.0,
    unit="meV",
    unique_name="TestGaussian1",
)
component2 = Lorentzian(
    display_name="TestLorentzian1",
    area=2.0,
    center=1.0,
    width=0.5,
    unit="meV",
    unique_name="TestLorentzian1",
)
model.add_component(component1)
model.add_component(component2)

model.normalize_area()
# EXPECT
x = np.linspace(-10000, 10000, 1000000)  # Lorentzians have long tails
result = model.evaluate(x)
numerical_area = simpson(result, x)

print(numerical_area)
```

```python
component_collection=ComponentCollection()

# Creating components
gaussian=Gaussian(display_name='Gaussian',width=0.5,area=1)
dho = DampedHarmonicOscillator(display_name='DHO',center=1.0,width=0.3,area=2.0)
lorentzian = Lorentzian(display_name='Lorentzian',center=-1.0,width=0.2,area=1.0)
polynomial = Polynomial(display_name='Polynomial',coefficients=[0.1, 0, 0.5])  # y=0.1+0.5*x^2

# Adding components to the component collection
component_collection.add_component(gaussian)
component_collection.add_component(dho)
component_collection.add_component(lorentzian)
component_collection.add_component(polynomial)

x=np.linspace(-2, 2, 100)

plt.figure()
y=component_collection.evaluate(x)
plt.plot(x, y, label='Component collection')

for component in component_collection.components:
    y = component.evaluate(x)
    plt.plot(x, y, label=component.display_name)

plt.legend()
plt.show()
```

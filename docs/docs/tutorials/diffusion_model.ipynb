# Diffusion Model
We support several standard models of diffusion. Here we show examples of Brownian Translational Diffusion, Jump Translational Diffusion, Chudley-Elliott Jump Diffusion and Hall-Ross Jump Diffusion. In each model the scattering is a Lorentzian for every $Q$, and the models differ in how the width $\Gamma(Q)$ depends on $Q$. For Brownian Translational Diffusion, the width is given by $\Gamma = D Q^2$, where $D$ is the diffusion coefficient (in m$^2$/s) and $Q$ is the momentum transfer.

Models with an elastic line and several Lorentzians have their own tutorials: [equivalent sites on a circle](equivalent_sites_circle.ipynb), [jump sites with a log-normal distribution](jump_sites_log_normal_distribution.ipynb), [the 3D Gaussian model](gaussian_model_3d.ipynb), [diffusion in a sphere](diffusion_in_sphere.ipynb) and [isotropic rotational diffusion](isotropic_rotational_diffusion.ipynb).

```python
import matplotlib.pyplot as plt
import numpy as np

import easydynamics as edyn

%matplotlib widget
```

## Brownian Translational Diffusion

### Model for different Q values

```python
# Create Brownian Translational Diffusion model
# and plot the model for different Q values.
# Q is in Angstrom^-1 and energy in meV.

Q = np.linspace(0.5, 2, 7)
energy = np.linspace(-2, 2, 501)
scale = 1.0
diffusion_coefficient = 2.4e-9  # m^2/s

diffusion_model = edyn.BrownianTranslationalDiffusion(
    display_name='DiffusionModel', scale=scale, diffusion_coefficient=diffusion_coefficient, Q=Q
)

component_collections = diffusion_model.get_component_collections()


cmap = plt.cm.jet
nQ = len(component_collections)
plt.figure()
for Q_index in range(len(component_collections)):
    color = cmap(Q_index / (nQ - 1))
    y = component_collections[Q_index].evaluate(energy)
    plt.plot(energy, y, label=f'Q={Q[Q_index]} Å^-1', color=color)

plt.legend()
plt.show()
plt.xlabel('Energy (meV)')
plt.ylabel('Intensity (arb. units)')
plt.title('Brownian Translational Diffusion Model')
```

### Width as a function of Q

```python
# Calculate and plot the half width at half maximum (HWHM) as function
# of Q
Q = np.linspace(0.1, 2, 101)
HWHM = diffusion_model.calculate_width(Q)
plt.figure()
plt.plot(Q, HWHM)
plt.xlabel('Q (Å$^{-1}$)')
plt.ylabel('HWHM (meV)')
plt.xlim(0, 2.5)
plt.ylim(0, max(HWHM) * 1.1)
plt.title('HWHM vs Q for Brownian Translational Diffusion')
```

## Jump Translational Diffusion
In Jump Translational Diffusion, the particle waits for a relaxation time $t$ between jumps. At low $Q$ the width follows the Brownian result, $\Gamma \approx \hbar D Q^2$, but at high $Q$ it levels off:

$$\Gamma(Q) = \frac{\hbar D Q^2}{1 + D t Q^2},$$

where $D$ is the diffusion coefficient (in m$^2$/s) and $t$ is the relaxation time (in ps).

### Model for different Q values

```python
# Create Jump Translational Diffusion model
# and plot the model for different Q values.
# Q is in Angstrom^-1 and energy in meV.

Q = np.linspace(0.5, 2, 7)
energy = np.linspace(-2, 2, 501)
scale = 1.0
diffusion_coefficient = 2.4e-9  # m^2/s
relaxation_time = 1.0  # ps

jump_diffusion_model = edyn.JumpTranslationalDiffusion(
    display_name='JumpDiffusionModel',
    scale=scale,
    diffusion_coefficient=diffusion_coefficient,
    relaxation_time=relaxation_time,
    Q=Q,
)

component_collections = jump_diffusion_model.get_component_collections()

cmap = plt.cm.jet
nQ = len(component_collections)
plt.figure()
for Q_index in range(len(component_collections)):
    color = cmap(Q_index / (nQ - 1))
    y = component_collections[Q_index].evaluate(energy)
    plt.plot(energy, y, label=f'Q={Q[Q_index]} Å^-1', color=color)

plt.legend()
plt.xlabel('Energy (meV)')
plt.ylabel('Intensity (arb. units)')
plt.title('Jump Translational Diffusion Model')
plt.show()
```

### Width as a function of Q

```python
# Calculate and plot the half width at half maximum (HWHM) as function
# of Q
Q = np.linspace(0.1, 2, 101)
HWHM = jump_diffusion_model.calculate_width(Q)
plt.figure()
plt.plot(Q, HWHM)
plt.xlabel('Q (Å$^{-1}$)')
plt.ylabel('HWHM (meV)')
plt.xlim(0, 2.5)
plt.ylim(0, max(HWHM) * 1.1)
plt.title('HWHM vs Q for Jump Translational Diffusion')
plt.show()
```

## Chudley-Elliott Jump Diffusion
The Chudley-Elliott model<sup>1</sup> describes a particle that sits on a site for a residence time $\tau$ and then jumps a fixed distance $l$. The width is

$$\Gamma(Q) = \frac{\hbar}{\tau}\left(1 - \frac{\sin(Ql)}{Ql}\right),$$

where $\tau$ is the residence time (in ps) and $l$ is the jump length (in Å). The width oscillates with $Q$ and approaches $\hbar/\tau$ at high $Q$.

### Model for different Q values

```python
# Create Chudley-Elliott Jump Diffusion model
# and plot the model for different Q values.
# Q is in Angstrom^-1 and energy in meV.

Q = np.linspace(0.5, 2, 7)
energy = np.linspace(-2, 2, 501)
scale = 1.0
residence_time = 1.0  # ps
jump_length = 1.5  # Å

chudley_elliott_model = edyn.ChudleyElliottJumpDiffusion(
    display_name='ChudleyElliottModel',
    scale=scale,
    residence_time=residence_time,
    jump_length=jump_length,
    Q=Q,
)

component_collections = chudley_elliott_model.get_component_collections()

cmap = plt.cm.jet
nQ = len(component_collections)
plt.figure()
for Q_index in range(len(component_collections)):
    color = cmap(Q_index / (nQ - 1))
    y = component_collections[Q_index].evaluate(energy)
    plt.plot(energy, y, label=f'Q={Q[Q_index]} Å^-1', color=color)

plt.legend()
plt.xlabel('Energy (meV)')
plt.ylabel('Intensity (arb. units)')
plt.title('Chudley-Elliott Jump Diffusion Model')
plt.show()
```

### Width as a function of Q

```python
# Calculate and plot the half width at half maximum (HWHM) as function
# of Q
Q = np.linspace(0.1, 4, 201)
HWHM = chudley_elliott_model.calculate_width(Q)
plt.figure()
plt.plot(Q, HWHM)
plt.xlabel('Q (Å$^{-1}$)')
plt.ylabel('HWHM (meV)')
plt.xlim(0, 4.2)
plt.ylim(0, max(HWHM) * 1.1)
plt.title('HWHM vs Q for Chudley-Elliott Jump Diffusion')
plt.show()
```

## Hall-Ross Jump Diffusion
The Hall-Ross model<sup>2</sup> describes a particle that sits on a site for a residence time $\tau$ and then jumps a random distance drawn from a Gaussian distribution. The width is

$$\Gamma(Q) = \frac{\hbar}{\tau}\left(1 - \exp\left(-\frac{Q^2 l^2}{2}\right)\right),$$

where $\tau$ is the residence time (in ps) and $l$ is the standard deviation of the jump-length distribution (in Å).<sup>3</sup> The mean-square jump length is $\langle r^2 \rangle = 3 l^2$. Unlike Chudley-Elliott, the width does not oscillate: it rises monotonically from $\hbar D Q^2$, with $D = l^2 / (2\tau)$, to $\hbar/\tau$.

### Model for different Q values

```python
# Create Hall-Ross Jump Diffusion model
# and plot the model for different Q values.
# Q is in Angstrom^-1 and energy in meV.

Q = np.linspace(0.5, 2, 7)
energy = np.linspace(-2, 2, 501)
scale = 1.0
residence_time = 1.0  # ps
jump_length = 1.5 / np.sqrt(3)  # Å, the standard deviation l

hall_ross_model = edyn.HallRossJumpDiffusion(
    display_name='HallRossModel',
    scale=scale,
    residence_time=residence_time,
    jump_length=jump_length,
    Q=Q,
)

component_collections = hall_ross_model.get_component_collections()

cmap = plt.cm.jet
nQ = len(component_collections)
plt.figure()
for Q_index in range(len(component_collections)):
    color = cmap(Q_index / (nQ - 1))
    y = component_collections[Q_index].evaluate(energy)
    plt.plot(energy, y, label=f'Q={Q[Q_index]} Å^-1', color=color)

plt.legend()
plt.xlabel('Energy (meV)')
plt.ylabel('Intensity (arb. units)')
plt.title('Hall-Ross Jump Diffusion Model')
plt.show()
```

### Width as a function of Q
The standard deviation above was chosen as $l = 1.5/\sqrt{3}$ Å, so that the mean-square jump length, $3 l^2 = 2.25$ Å$^2$, equals that of the Chudley-Elliott model above with its fixed jump of 1.5 Å. Both models then have the same diffusion coefficient and agree at low $Q$; at high $Q$ the Hall-Ross width approaches $\hbar/\tau$ without oscillating.

```python
# Calculate and plot the half width at half maximum (HWHM) as function
# of Q, together with the Chudley-Elliott model with the same mean-square jump length
Q = np.linspace(0.1, 4, 201)
HWHM = hall_ross_model.calculate_width(Q)
plt.figure()
plt.plot(Q, HWHM, label='Hall-Ross')
plt.plot(Q, chudley_elliott_model.calculate_width(Q), linestyle='dashed', label='Chudley-Elliott')
plt.xlabel('Q (Å$^{-1}$)')
plt.ylabel('HWHM (meV)')
plt.xlim(0, 4.2)
plt.ylim(0, max(HWHM) * 1.2)
plt.title('HWHM vs Q for Hall-Ross Jump Diffusion')
plt.legend()
plt.show()
```
